# Figure 1

**A summary of the workflows supported in Mycol**

- **Is** - Inkscape artwork, not a plot; edit `assets/Figure_1.svg` to change it
- **Does** - copies the master `.svg` and its 500 dpi `.png` into `output/`
- **Note** - the artwork reads "different celltypes"; should be "cell types"
- **Kernel** - `mycol_colonies_env`, run top to bottom


```python
import shutil
from pathlib import Path

from IPython.display import display
from PIL import Image

OUTPUT = Path("output")
OUTPUT.mkdir(exist_ok=True)
for name in ("Figure_1.svg", "Figure_1.png"):
    out = shutil.copy2(f"assets/{name}", OUTPUT / name)
    print(f"wrote {out}  {Path(out).stat().st_size:,} B")

im = Image.open(OUTPUT / "Figure_1.png")
print(f"\nFigure 1   {im.width} x {im.height} px")
im.thumbnail((760, 760))
display(im.convert("RGB"))
```
